```python
# PROJECT 1: DATA CLEANING & PREPARATION
# ============================================

import pandas as pd
import numpy as np

# --------------------------------------------
# 1. RAW DATASET
# --------------------------------------------

data = {
    "Name": ["Rahul", "Priya", "Aman", "Rahul", "Neha", "Rohit", None],
    "Age": [21, 22, None, 21, 25, 150, 23],
    "Salary": [25000, 30000, 28000, 25000, None, 50000, 32000],
    "City": ["Delhi", "Mumbai", "Delhi", "Delhi", "Agra", "Mumbai", "Agra"],
    "Join_Date": [
        "2025-01-10",
        "2025-02-15",
        "2025-03-20",
        "2025-01-10",
        "2025-04-05",
        "wrong_date",
        "2025-05-10"
    ]
}

df = pd.DataFrame(data)

print("========== RAW DATASET ==========")
display(df)


# --------------------------------------------
# 2. BASIC INFORMATION
# --------------------------------------------

print("\n========== DATASET INFORMATION ==========")
print("Rows and Columns:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())


# --------------------------------------------
# 3. REMOVE DUPLICATES
# --------------------------------------------

print("\n========== DUPLICATE CHECK ==========")
print("Duplicates before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after:", df.duplicated().sum())


# --------------------------------------------
# 4. HANDLE MISSING NAME
# --------------------------------------------

df["Name"] = df["Name"].fillna("Unknown")


# --------------------------------------------
# 5. CORRECT INCORRECT AGE
# --------------------------------------------

# Age above 100 is considered incorrect
df.loc[df["Age"] > 100, "Age"] = np.nan

# Fill missing Age using median
df["Age"] = df["Age"].fillna(df["Age"].median())
# --------------------------------------------
# 6. HANDLE MISSING SALARY
# --------------------------------------------

df["Salary"] = df["Salary"].fillna(df["Salary"].median())


# --------------------------------------------
# 7. CORRECT DATE FORMAT
# --------------------------------------------

df["Join_Date"] = pd.to_datetime(
    df["Join_Date"],
    errors="coerce"
)

# Fill invalid date using most common date
df["Join_Date"] = df["Join_Date"].fillna(
    df["Join_Date"].mode()[0]
)


# --------------------------------------------
# 8. FINAL CLEAN DATASET
# --------------------------------------------

print("\n========== CLEANED DATASET ==========")
display(df)


# --------------------------------------------
# 9. FINAL CHECK
# --------------------------------------------

print("\n========== FINAL CHECK ==========")

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)


# ------
# 10. BASIC DATA ANALYSIS
# --------------------------------------------

print("\n========== BASIC ANALYSIS ==========")

print("\nAverage Age:", round(df["Age"].mean(), 2))

print("Average Salary:", round(df["Salary"].mean(), 2))

print("\nCity-wise Employee Count:")
print(df["City"].value_counts())


# --------------------------------------------
# 11. SAVE CLEANED DATASET
# --------------------------------------------

df.to_csv("cleaned_dataset.csv", index=False)

print("\n============================================")
print("Cleaned dataset saved as cleaned_dataset.csv")
print("Project 1 Completed Successfully!")
print("============================================")
```
